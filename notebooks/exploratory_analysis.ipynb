# DataPulse: Exploratory Data Analysis & Student Performance Modeling
**Institution**: Universiapolis International University of Agadir (UIA)
**Dataset**: UCI Student Performance (ID: 320)
**Objective**: Descriptive analysis, feature engineering, and academic risk classification.

```python
import pandas as pd
import numpy as np

# 1. Load Raw Datasets
df_mat = pd.read_csv('../data/raw/student-mat.csv', sep=';')
df_por = pd.read_csv('../data/raw/student-por.csv', sep=';')

print(f"Mathematics Dataset: {df_mat.shape[0]} rows, {df_mat.shape[1]} columns")
print(f"Portuguese Dataset:  {df_por.shape[0]} rows, {df_por.shape[1]} columns")
```

```python
# 2. Summary of Target Variable (Final Grade G3)
print("Mathematics G3 Summary:")
print(df_mat['G3'].describe())
print("\nPortuguese G3 Summary:")
print(df_por['G3'].describe())
```

```python
# 3. Feature Engineering & Entity Linking
df_mat['course'] = 'MAT'
df_por['course'] = 'POR'
all_data = pd.concat([df_mat, df_por], ignore_index=True)

# Compute Derived Features
all_data['parent_edu_avg'] = (all_data['Medu'] + all_data['Fedu']) / 2.0
all_data['alcohol_index'] = (all_data['Dalc'] * 5.0 + all_data['Walc'] * 2.0) / 7.0
all_data['study_to_social'] = all_data['studytime'] / (all_data['goout'] + all_data['freetime'] + 0.1)
all_data['academic_risk'] = (all_data['G3'] < 10).astype(int)

print("Engineered dataset shape:", all_data.shape)
print(f"Overall Academic Risk Rate: {(all_data['academic_risk'].mean() * 100):.1f}%")
```

```python
# 4. Correlation with Final Grade
num_cols = ['G3', 'G1', 'G2', 'studytime', 'failures', 'absences', 'parent_edu_avg', 'alcohol_index']
corrs = all_data[num_cols].corr()['G3'].sort_values(ascending=False)
print("Correlation with Final Grade (G3):")
print(corrs)
```
